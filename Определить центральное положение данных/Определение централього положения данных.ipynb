### Условие задачи

Необходимо визуально оценить центральное положение данных о зарплатах сотрудников в компании, определить среднюю и медианную зарплату в компании.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Загрузка данных из CSV-файла
data = pd.read_csv('salary_data.csv')
data.head()
```

```python
# Рассчет среднего значения и медианы
mean_salary = data['salary'].mean()
median_salary = data['salary'].median()
```

```python
# Визуализация распределения зарплат
plt.figure(figsize=(8, 6))
plt.hist(data['salary'], bins=10, edgecolor='black')
plt.xlabel('Зарплата')
plt.ylabel('Частота')
plt.title('Распределение зарплат')
plt.show()
```

```python

# Вывод результатов
print("Средняя зарплата: ", mean_salary)
print("Медиана зарплаты: ", median_salary)
```

## Расчитать корреляцию и построить гистограмму

Необходимо исследовать взаимосвязь между двумя переменными в наборе данных: цена товара и объем продаж. Задача: рассчитать корреляцию между этими переменными и построить диаграмму рассеяния для визуализации результатов.

```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import pearsonr
```

```python
# Загрузка данных из файла CSV
data = pd.read_csv('data.csv')

data.head()
```

```python
# Выделение двух переменных для исследования
Price = data['Price']
Value = data['Value']
```

```python
# Расчет корреляции
correlation = pearsonr(Price, Value)
correlation
```

```python
# Построение диаграммы рассеяния
plt.scatter(Price, Value)
plt.xlabel('Price')
plt.ylabel('Value')
plt.title('Scatter Plot')
plt.show()
```

```python
# Выводы анализа
print("Корреляция между ценой и объемом продаж:", correlation)
```
